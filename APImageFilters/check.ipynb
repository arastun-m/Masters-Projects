```python
from PIL import Image
import matplotlib.pyplot as plt

# Specify image paths
input_image_path = "Images/gracehopper.png"  
output_image_path = "output.png"      
hsl_image_path = "output_hsl.png"
hsv_image_path = "output_hsv.png"

# Load images and convert to grayscale
input_img = Image.open(input_image_path).convert("L")
output_img = Image.open(output_image_path).convert("L")
hsl_image = Image.open(hsl_image_path).convert("L")
hsv_image = Image.open(hsv_image_path).convert("L")

# Get histogram data
input_hist = input_img.histogram()
output_hist = output_img.histogram()
hsl_hist = hsl_image.histogram()
hsv_hist = hsv_image.histogram()

# Create a 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot histograms
axes[0, 0].bar(range(256), input_hist, color='gray', width=1.0)
axes[0, 0].set_title("Original Image Histogram")
axes[0, 0].set_xlabel("Pixel Intensity")
axes[0, 0].set_ylabel("Frequency")
axes[0, 0].set_xlim([0, 255])

axes[0, 1].bar(range(256), output_hist, color='gray', width=1.0)
axes[0, 1].set_title("Standard Output Histogram")
axes[0, 1].set_xlabel("Pixel Intensity")
axes[0, 1].set_ylabel("Frequency")
axes[0, 1].set_xlim([0, 255])

axes[1, 0].bar(range(256), hsl_hist, color='gray', width=1.0)
axes[1, 0].set_title("HSL Luminance Equalization Histogram")
axes[1, 0].set_xlabel("Pixel Intensity")
axes[1, 0].set_ylabel("Frequency")
axes[1, 0].set_xlim([0, 255])

axes[1, 1].bar(range(256), hsv_hist, color='gray', width=1.0)
axes[1, 1].set_title("HSV Value Equalization Histogram")
axes[1, 1].set_xlabel("Pixel Intensity")
axes[1, 1].set_ylabel("Frequency")
axes[1, 1].set_xlim([0, 255])

plt.tight_layout()
plt.show()
```
